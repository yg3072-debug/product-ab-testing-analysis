# E-commerce Checkout A/B Test  
## Experiment Design

### 1. Business Context

The e-commerce platform observed substantial user drop-off after customers entered the checkout process. The existing checkout experience requires users to complete shipping information, payment details, and order review across multiple steps.

The product team developed a simplified single-page checkout experience intended to reduce friction and make the purchase process faster and easier to complete.

This experiment evaluates whether the new checkout flow increases completed purchases without negatively affecting payment reliability, order quality, or the overall customer experience.

---

### 2. Product Change

**Control:** Existing multi-step checkout flow.

**Treatment:** Simplified single-page checkout flow combining shipping information, payment details, and order review on one page.

---

### 3. Experiment Objective

The objective of the experiment is to determine whether the simplified checkout flow increases purchase conversion among users who reach checkout.

The experiment should also assess whether any conversion improvement is accompanied by higher payment failure, checkout errors, refunds, or cancellations.

---

### 4. Hypotheses

Let:

- \(p_c\) represent the checkout purchase conversion rate for the control group.
- \(p_t\) represent the checkout purchase conversion rate for the treatment group.

**Null hypothesis**

$$
H_0: p_t = p_c
$$

**Alternative hypothesis**

$$
H_1: p_t \neq p_c
$$

The business hypothesis is that simplifying the checkout experience will reduce user friction and increase the proportion of checkout-exposed users who complete a purchase.

A two-sided statistical test will be used because the new experience may either improve or reduce conversion.

---

### 5. Randomization Unit

The experiment will be randomized at the user level.

Each user should be assigned to exactly one experiment group and remain in that group throughout the experiment.

User-level randomization prevents the same user from experiencing both checkout versions across different sessions and aligns the analysis unit with the randomization unit.

---

### 6. Experiment Population

The primary analysis population includes users who:

- reached the checkout page during the experiment assignment window;
- received a valid control or treatment assignment;
- had a valid user identifier;
- were exposed to exactly one experiment group;
- were not internal employees, test accounts, or identified bot traffic; and
- had a complete 24-hour conversion observation window.

Users who were assigned to the experiment but never reached checkout will not be included in the primary exposed-user analysis.

---

### 7. Experiment Exposure

A user is considered exposed when the user records the first valid checkout page view after experiment assignment.

The first valid exposure timestamp will be used as the beginning of the user's conversion attribution window.

---

### 8. Primary Metric

#### Checkout Purchase Conversion Rate

$$
\text{Checkout Purchase Conversion Rate}
=
\frac{\text{Unique exposed users completing at least one purchase}}
{\text{Unique checkout-exposed users}}
$$

A user is counted as converted if the user completes at least one successful purchase within 24 hours after the first valid checkout exposure.

The metric will be calculated at the user level, with one binary outcome per user:

- `purchased = 1`: the user completed at least one attributed purchase;
- `purchased = 0`: the user did not complete an attributed purchase.

---

### 9. Secondary Metrics

#### Checkout Completion Time

Median time between first checkout exposure and completed purchase among converted users.

#### Payment Success Rate

$$
\text{Payment Success Rate}
=
\frac{\text{Users with a successful payment}}
{\text{Users who attempted payment}}
$$

#### Revenue per Exposed User

$$
\text{Revenue per Exposed User}
=
\frac{\text{Total completed purchase revenue}}
{\text{Unique checkout-exposed users}}
$$

These metrics help explain whether the new flow improves checkout speed, payment completion, and overall commercial value.

---

### 10. Guardrail Metrics

#### Payment Failure Rate

$$
\text{Payment Failure Rate}
=
\frac{\text{Users experiencing a payment failure}}
{\text{Users who attempted payment}}
$$

#### Checkout Error Rate

$$
\text{Checkout Error Rate}
=
\frac{\text{Exposed users experiencing at least one checkout error}}
{\text{Checkout-exposed users}}
$$

#### Refund or Cancellation Rate

$$
\text{Refund or Cancellation Rate}
=
\frac{\text{Purchasers whose orders were refunded or cancelled}}
{\text{Purchasers with a complete follow-up window}}
$$

Guardrail metrics will be used to determine whether conversion improvements come at the cost of payment reliability, technical stability, or order quality.

---

### 11. Metric Definition Table

| Metric | Role | Numerator | Denominator | Analysis Unit | Observation Window |
|---|---|---|---|---|---|
| Checkout purchase conversion | Primary | Users completing at least one purchase | Checkout-exposed users | User | 24 hours after exposure |
| Checkout completion time | Secondary | Not applicable | Not applicable | Converted user | Exposure to purchase |
| Payment success rate | Secondary | Users with successful payment | Users attempting payment | User | 24 hours after exposure |
| Revenue per exposed user | Secondary | Completed purchase revenue | Checkout-exposed users | User | 24 hours after exposure |
| Payment failure rate | Guardrail | Users with payment failure | Users attempting payment | User | 24 hours after exposure |
| Checkout error rate | Guardrail | Users with a checkout error | Checkout-exposed users | User | Checkout session |
| Refund or cancellation rate | Guardrail | Refunded or cancelled purchasers | Purchasers with complete follow-up | User or order | 7 days after purchase |

---

### 12. Planned Segment Analysis

Exploratory segment analysis will evaluate experiment results by:

- device at experiment exposure;
- traffic source at experiment exposure; and
- new versus returning user status.

Segment results will be treated as exploratory unless the segments and hypotheses are formally specified before analyzing the outcomes.

---

### 13. Experiment Window

- **Assignment window:** July 1, 2026 through July 14, 2026
- **Primary conversion window:** 24 hours after first valid checkout exposure
- **Refund and cancellation follow-up:** 7 days after purchase

Users without a complete outcome observation window will be excluded from the corresponding mature analysis.

---

### 14. Decision Framework

The treatment will be considered a strong launch candidate if:

1. checkout purchase conversion improves with statistical significance;
2. the absolute improvement is at least 1.0 percentage point;
3. payment failure and checkout error rates do not materially worsen;
4. revenue per exposed user remains neutral or positive;
5. refund and cancellation rates remain within acceptable limits; and
6. the overall result is not driven entirely by a small or anomalous segment.

Statistical significance alone will not be treated as sufficient evidence for launch. The decision will consider effect size, business value, data quality, and guardrail performance.

---

### 15. Key Analytical Risks

The analysis must account for the following risks:

- users assigned to multiple experiment groups;
- duplicate experiment exposure events;
- purchases occurring before experiment exposure;
- incomplete 24-hour conversion windows;
- incomplete refund and cancellation follow-up;
- event-level joins that duplicate users or orders;
- bot, employee, or test-account traffic;
- inconsistent device or traffic-source attribution;
- outcome-based definitions of user segments;
- repeated segment testing and false-positive findings.

These issues will be evaluated during SQL data validation and Python statistical analysis.

---

## Analyst Reflection

### Question 1

**Why should the experiment denominator include checkout-exposed users rather than all active website users?**

The denominator should include checkout-exposed users because only users who reached the checkout page had an opportunity to experience the product change. Including all active website users would add users who were never exposed to either checkout version and could not have been affected by the treatment. This would dilute the measured treatment effect and mix upstream behaviors, such as product browsing and add-to-cart activity, into a checkout-specific experiment.

### Question 2

**Why should the analysis use one row per user instead of one row per event?**

The analysis should use one row per user because the experiment is randomized at the user level. A single user may generate multiple checkout, payment, or purchase events, so event-level analysis could give more weight to highly active users and violate the independence assumption of the statistical test. Aggregating the data to one binary purchase outcome per user aligns the analysis unit with the randomization unit and prevents duplicate counting.

### Question 3

**Why is a 24-hour attribution window necessary?**

A 24-hour attribution window defines which purchases can reasonably be attributed to the checkout experience. Without a fixed window, purchases occurring several days later might be incorrectly counted as experiment outcomes, and users entering the experiment earlier would have more time to convert than users entering later. The window creates a consistent observation period and reduces bias caused by unequal follow-up time.

### Question 4

**Why might a statistically significant conversion increase still be insufficient for launching the treatment?**

A statistically significant result does not necessarily mean that the improvement is large enough to create meaningful business value. A very small conversion increase may become statistically significant with a large sample but may not justify the development, maintenance, or operational cost of the new checkout flow. The treatment should also be evaluated against guardrail metrics, because higher conversion would not support a launch if it caused more payment failures, checkout errors, refunds, cancellations, or lower revenue per exposed user.
