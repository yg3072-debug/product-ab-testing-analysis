```python
import pandas as pd
import numpy as np

np.random.seed(42)

users = pd.DataFrame({
    "user_id": range(1, 1001),
    "signup_date": pd.to_datetime("2026-01-01") + pd.to_timedelta(np.random.randint(0, 180, 1000), unit="D"),
    "country": np.random.choice(["US", "Canada", "Mexico", "China"], 1000, p=[0.6, 0.15, 0.15, 0.1]),
    "device": np.random.choice(["mobile", "desktop", "tablet"], 1000, p=[0.6, 0.35, 0.05])
})

orders = pd.DataFrame({
    "order_id": range(1, 1501),
    "user_id": np.random.choice(users["user_id"], 1500),
    "order_date": pd.to_datetime("2026-03-01") + pd.to_timedelta(np.random.randint(0, 90, 1500), unit="D"),
    "amount": np.round(np.random.exponential(scale=50, size=1500), 2),
    "status": np.random.choice(["completed", "cancelled", "refunded"], 1500, p=[0.85, 0.1, 0.05])
})

events = pd.DataFrame({
    "event_id": range(1, 5001),
    "user_id": np.random.choice(users["user_id"], 5000),
    "event_date": pd.to_datetime("2026-03-01") + pd.to_timedelta(np.random.randint(0, 90, 5000), unit="D"),
    "event_type": np.random.choice(["view_product", "add_to_cart", "checkout", "purchase"], 5000, p=[0.55, 0.25, 0.12, 0.08]),
    "traffic_source": np.random.choice(["organic", "paid_search", "social", "email"], 5000)
})

users.to_csv("D:/JNP/analytics 30days July/AB testing/data/users.csv", index=False)
orders.to_csv("D:/JNP/analytics 30days July/AB testing/data/orders.csv", index=False)
events.to_csv("D:/JNP/analytics 30days July/AB testing/data/events.csv", index=False)
```
